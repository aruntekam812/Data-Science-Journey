# 5. Multi-Plot / Advanced EDA

# FacetGrid()

```python
df1
```

```python

g = sns.FacetGrid(df1, col='day',hue='sex')
g.map(plt.scatter, 'total_bill','tip').add_legend()
sns.set_context('poster',font_scale=1)
plt.show()
```

# Catplot()

```python
sns.catplot(x='total_bill',y='tip',data=df1,hue= 'sex',palette='Accent',kind='strip')
plt.show()
```

# relplot()

```python
sns.relplot(x='total_bill',y='tip',data=df1,hue= 'sex')
plt.show()
```

# 6. Styling & Themes

# set_style

```python
sns.set_style('darkgrid')
sns.barplot(x='day',y='total_bill',data=df1)
plt.show()
```

# set_context()

```python
sns.set_style('white')
sns.histplot(x='total_bill',data=df1)
sns.set_context('poster',font_scale=1)
plt.show()
```

# despine()

```python
sns.set_style('white')
sns.histplot(x='total_bill',data=df1,hue='sex',palette="cool")
sns.despine()
plt.show()
```
