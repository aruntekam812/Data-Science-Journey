#  CREATE CHARTS


```python
import matplotlib.pyplot as plt
```

# 1.LINE GRAPH (For looking trend)

```python
x = [1,2,3,4,5]
y=[10,20,25,35,45]

plt.plot(x,y)
plt.show()
```

**Customize chart**

```python
plt.figure(figsize=(4,2))                   #for reduce size

plt.plot(x,y, color='green',marker='o',linestyle='dashed',linewidth=2,markersize=12)

plt.title('market pricing')                  #add title name

plt.xlabel('months')                         # x-axis name

plt.ylabel('sales')                           # y-axis name


plt.plot(x,y,label='sales')                  #show multiple line label
plt.legend()

plt.grid() #                               add grid line in background

plt.show()
```

**legend**

```python
plt.figure(figsize=(6,4))
x= [1,2,3,4]
y1=[5,6,7,8]
y2=[9,10,11,12]

plt.plot(x,y1,label='sales 2024')
plt.plot(x,y2,label='sales 2025')
plt.title('yo sales')
plt.xlabel('month')
plt.ylabel('sales')

plt.legend()
plt.show()
```

# 3️⃣ Category Plots

# plt.bar()

```python
plt.figure(figsize=(4,6))
x =['A','B','C']
y=[80,70,90]

plt.bar(x,y)
plt.show()
```

# plt.barh()

# 4️⃣ Figure Control

# plt.subplot()

```python
y1=[5,6,7,8]
y2=[9,10,11,12]

x =['A','B','C','D','E']
y = [10,20,55,35,45]

plt.figure(figsize=(6,4))

plt.subplot(1,2,1) #ROW ,COLUMN,POSITION
# bar chart
plt.bar(y1,y2)
plt.title('bar user')
plt.xlabel('bar month')
plt.ylabel('bar sales')

plt.subplot(1,2,2)
#scatter chart
plt.scatter(x,y)
plt.title('scatter bolte')
plt.xlabel('month')
plt.ylabel('sales')
plt.show()
```

# 5️⃣ Styling

# plt.style.use()

```python
import matplotlib.pyplot as plt
import numpy as np

# Apply the ggplot style
plt.style.use('ggplot')

# Create a sample plot
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x))
plt.title("Sine Wave in ggplot Style")
plt.show()
```

#  6️⃣ Advanced Functions

# plt.pie()


```python
x =['A','B','C','D','E']
y = [10,20,55,35,45]
plt.pie(y,labels=x,autopct='%1.1f%%',startangle=90)
plt.show()
```

# plt.annotate()

```python
plt.annotate('peak',(45,55))   # ADD TEXT ON GRAPH


```

# plt.axvline()  or plt.axhline()

```python
x =['A','B','C','D','E']
y = [10,20,55,35,45]
plt.plot(x,y)
plt.axvline(x=2)
plt.axhline(y=30)
plt.show()
```

# 7️⃣ Save Output

```python
x =['A','B','C','D','E']
y = [10,20,55,35,45]
plt.pie(y,labels=x,autopct='%1.1f%%',startangle=90)

plt.savefig("grap.png") # SAVE GRAPH OR IMAGE INTO PDF FORM
plt.show()
```
