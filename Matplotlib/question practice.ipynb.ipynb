```python
import pandas as pd
import matplotlib.pyplot as plt


# Dummy data for all 25 exercises
data = {
    'Student': ['Amit', 'Priya', 'Rohan', 'Sneha', 'Vikram'],
    'Math_Marks': [85, 92, 78, 95, 64],
    'Science_Marks': [88, 89, 82, 91, 70],
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May'],
    'Company_A_Sales': [120, 150, 180, 220, 200],
    'Company_B_Sales': [90, 130, 170, 190, 240],
    'Temperature': [15, 18, 25, 32, 38],
    'City': ['Mumbai', 'Delhi', 'Bangalore', 'Hyderabad', 'Ahmedabad'],
    'Population_Millions': [21.0, 19.8, 12.3, 10.0, 8.2],
    'Experience_Years': [1, 3, 5, 8, 10],
    'Salary_Thousands': [40, 55, 80, 120, 150],
    'Mobile_Brand': ['Samsung', 'Apple', 'Xiaomi', 'Vivo', 'Others'],
    'Market_Share': [30, 25, 20, 15, 10]
}

df = pd.DataFrame(data)
df.to_csv('sample_data.csv', index=False)
print("CSV file 'sample_data.csv' successfully created!")
```

# 1️⃣ Basic Line Plot

Student marks ka line graph banao.

```python
plt.plot(df['Student'],df['Science_Marks'])
plt.title("Basic Line Plot - Student Marks")
df['Math_Marks'].sum()
plt.figure()
plt.plot(df['Student'], df['Math_Marks'])
plt.title("Basic Line Plot - Student Marks")
plt.show()
```

# 2️⃣ Multiple Line Plot

Math aur Science marks ko same graph me compare karo.

```python
plt.figure()
plt.plot(df['Student']	,df['Math_Marks'],label='math')
plt.plot(df['Student']	,df['Science_Marks'],label='science')
plt.legend()
plt.title('math vs science marks')
plt.show()
```

#  3️⃣ Add Title & Labels

Sales data ka graph banao with:

title
xlabel
ylabel

```python
plt.plot(df['Company_A_Sales'])
plt.title('company a sale chart')
plt.xlabel('month')
plt.ylabel('sales')
plt.show()
df.query("Company_A_Sales>150")[['Month','Company_A_Sales']]
```

# 4️⃣ Add Grid

Monthly temperature graph me grid add karo.

```python
plt.plot(df['Temperature'])
plt.grid()
plt.show()
```

# 5️⃣ Add Legend

2 company sales trend ko legend ke sath show karo.  

```python
plt.plot(df['Company_A_Sales'],df['Company_B_Sales'],label='A')
plt.plot(df['Company_A_Sales'],df['Company_B_Sales'],label='B')
plt.legend()
plt.figure(figsize=(10,5))
plt.show()
```

# 7️⃣ Bar Chart

Students ke marks ka bar chart banao.

```python
plt.bar(df['Student'],df['Math_Marks'])
plt.title('Students ke marks maths')
plt.figure()
plt.bar(df['Student'],df['Science_Marks'])
plt.title('Students ke marks science')
plt.show()
```

# 8️⃣ Horizontal Bar Chart

Top 5 cities population ka horizontal chart banao.

```python
df.sort_values('City',ascending=False)
plt.figure(figsize=(8,3))
plt.barh(df['City'],df['Population_Millions'])
plt.title("Top 5 Cities Population")

plt.show()

```

# 9️⃣ Histogram

Random marks ka histogram banao.

👉 bins parameter use karo.

```python
import numpy as np
RM = np.random.randint(1,50,100)
plt.hist(RM,bins=10,edgecolor='black')
plt.title("Histogram of Random Marks")
plt.xlabel("Marks Range")
plt.ylabel("Frequency")
plt.show()

```

# 🔟 Scatter Plot

Experience vs Salary scatter plot banao.
