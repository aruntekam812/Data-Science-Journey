# Joining and Spliting

```python
arr= np.array([[1,2,3],[4,5,6],[7,1,0]])
```

```python
arr
```

```python
arr2=([[1, 2, 3],
       [4, 55, 6],
       [7, 1, 10]])
```

***Simple join***

```python
np.concatenate([arr,arr2])
```

***Vertical Join***

```python
np.vstack([arr,arr2])
```

*arr,arr* itself join

```python
np.vstack([arr,arr])
```

***Horizontal Join***

```python
np.hstack([arr,arr2])
```

```python
np.hstack([arr2,arr])
```

*arr,arr itself join*

```python
np.hstack([arr,arr])
```

**Split**

*using ravel() method**

```python
arr.ravel()
```

```python
np.split(arr,3)
```

# Linear Algebra

```python
arr= np.array([[1,2,3],[4,5,6],[7,1,0]])
```

```python
arr
```

```python
arr2=([[1, 2, 3],
       [4, 55, 6],
       [7, 1, 10]])
```

**Dot product**

```python
np.dot(arr,arr2)
```

**Matmul**

```python
np.matmul(arr,arr2)
```

**Inverse**

```python
np.linalg.inv(arr)
```

**Determinant**

```python
np.linalg.det(arr)
```

**Eigen value**

```python
np.linalg.eig(arr)
```
