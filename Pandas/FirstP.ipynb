# Merge and joint

```python
df1 = {
    'name':['Arun','ankit'],
     'age' : ['10','30']
}
```

```python
df2 = {
    'name':['Eren','mikasa'],
     'age2' : ['20','40']
}
```

```python
df1=pd.DataFrame(df1)
```

```python
df2=pd.DataFrame(df2)
```

```python
df1
```

```python
df2
```

**Joint tables**

```python
pd.merge(df1,df2)
```

**combine data verical**

```python
pd.concat([df1,df2])
```

**combine data horizontal**

```python
pd.concat([df1,df2],axis=1)

```

**Merge**

```python
pd.merge(df1, df2, how='inner',on='name')
```
