# LEVEL 3 — Real Analysis Questions (31–40)


**average score naam ka column banao.**

```python
df['avg score']=df[['math score', 'reading score', 'writing score']].mean(axis=1)
```

```python
df
```

**Highest average score wala student find karo.**

```python
df.sort_values('avg score')[['gender','avg score']].max()
```

**Lowest average score wala student find karo.**

```python
df.sort_values('avg score')[['gender','avg score']].min()
```

**Pass/fail column banao:
average >= 40 → Pass
otherwise Fail**

```python
df['pass/fail']=np.where(df['math score'] >=40,'pass','fail')

    
```

```python
df
```

**Total pass students count karo.**

```python
df.query("`pass/fail`=='pass'")[['pass/fail']].count()
```

**Total fail students count karo.**

```python
df.query("`pass/fail`=='fail'")[['pass/fail']].count()
```

**Kaunsi subject me highest average hai?**

```python
df.query("`math score` and `reading score` and `writing score`")[['math score','reading score','writing score']].mean().idxmax()
```

**Math vs reading correlation nikalo.**

```python
df[['math score','reading score']].corr()
```
